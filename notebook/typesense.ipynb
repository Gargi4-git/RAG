### LANGCHAIN + TYPESENSE + GROQ LLM + RAG APPLICATION

```python
from langchain.community.document_loaders import TextLoader
from langchian.commmunity.vectorstores import Typesense
from langchain.text_splitter import CharacterTextSplitter
from langchain.embeddings import HuggingFaceEmbeddings
from langchain_groq import ChatGroq
```

```python
import os
os.environ["GROQ_API_KEY"] = "YOUR_GROQ_API_KEY"
```

```python
loader = TextLoader("C:\\Users\\gargi\\Documents\\rag\\data\\text_files\\sample1.txt", encoding="utf-8")
documents = loader.load()
text_splitter = CharacterTextSplitter(chunk_size=1000, chunk_overlap=0)
docs = text_splitter.split_documents(documents)

embeddings = HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2")
```

```python
docsearch =Typesense.from_documents(
    docs, 
    embeddings, 
    typesense_client={
        "host": "s14hpbvkjq8ziwndp-1.a1.typesense.net",
        "port": "443",  
        "protocol": "https",
        "api_key": "T5Eem37DrfXTZSJjJGGI8oHt5aAa8jX2",  
        "typesense_collection": "books"
    })
```
